# Unemplyment study

```python
import pandas as pd
import matplotlib.pylab as plt

data = pd.read_csv('tepsr_wc170_linear.csv')
data.sample()
```

```python
romania = data[data.geo == 'Romania'] [['age', 'OBS_VALUE', 'TIME_PERIOD']]
young = romania[romania.age == 'From 15 to 24 years']

plt.figure(figsize=(15, 8))
plt.plot(young.TIME_PERIOD, young.OBS_VALUE)
plt.xticks(young.TIME_PERIOD)
plt.xlabel("Year")
plt.ylabel("Unemployment rate (as a % of pop)")
plt.title("Unemployment rate from 15 to 24 years (Romania)")
```

let's visualise all the age groups

```python
from functools import cmp_to_key
def compare_func(a, b):
    return int(a[11:13]) - int(b[11:13])

age_groups = sorted(set(romania.age), key=cmp_to_key(compare_func))
age_groups.remove('From 15 to 74 years')
plt.figure(figsize=(15, 8))

for age_group in age_groups:
    group = romania[romania.age == age_group]
    plt.plot(group.TIME_PERIOD, group.OBS_VALUE, label=age_group)
    
plt.xticks(young.TIME_PERIOD)
plt.xlabel("Year")
plt.ylabel("Unemployment rate (as a % of pop)")
plt.title("Unemployment rate in Romania by age group")
plt.legend()
```
